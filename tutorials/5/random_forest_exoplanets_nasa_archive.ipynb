```python
# Core scientific Python stack
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from urllib.parse import quote

from sklearn.model_selection import train_test_split, StratifiedKFold, cross_validate
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
from sklearn.impute import SimpleImputer
from sklearn.metrics import (
    classification_report,
    confusion_matrix,
    ConfusionMatrixDisplay,
    balanced_accuracy_score,
    f1_score
)
from sklearn.inspection import permutation_importance
from sklearn.ensemble import RandomForestClassifier

pd.set_option("display.max_columns", 200)
pd.set_option("display.width", 200)
```

## 1. Retrieve real data from the NASA Exoplanet Archive

We use the **`pscomppars`** table, which provides one row per planet and is often the most convenient table for statistical work.

### Why `pscomppars`?
- one row per planet,
- reasonably complete for population-level analysis,
- includes planet + host star parameters in one record.

We will request only a selected subset of columns to keep the exercise focused.

```python
BASE = "https://exoplanetarchive.ipac.caltech.edu/TAP/sync"

query = '''
select
    pl_name,
    hostname,
    discoverymethod,
    disc_year,
    pl_orbper,
    pl_orbsmax,
    pl_orbeccen,
    pl_bmasse,
    pl_dens,
    pl_eqt,
    pl_insol,
    pl_rade,
    st_teff,
    st_rad,
    st_mass,
    st_met,
    st_logg,
    sy_dist,
    sy_vmag,
    sy_kmag
from pscomppars
where pl_rade is not null
'''

url = f"{BASE}?query={quote(query)}&format=csv"
print(url)
```

```python
df = pd.read_csv(url)
print("Shape:", df.shape)
df.head()
```

## 3. Define the target classes from planetary radius

We create a 3-class target:
- rocky
- sub_neptune
- giant

```python
def radius_to_class(r):
    if pd.isna(r):
        return np.nan
    if r < 1.8:
        return "rocky"
    elif r < 4.0:
        return "sub_neptune"
    else:
        return "giant"

df["planet_class"] = df["pl_rade"].apply(radius_to_class)
df["planet_class"].value_counts(dropna=False)
```

```python
ax = df["planet_class"].value_counts().plot(kind="bar", figsize=(8,4))
ax.set_title("Target class counts")
ax.set_ylabel("Count")
plt.show()
```

The classes are reasonably balanced

## 4. Choose input features

We **exclude**
- `pl_rade` because it defines the target,
- names/identifiers because they do not represent physics,
- host names because they are identifiers rather than predictive variables.

We keep a compact physically interpretable set of orbital, planetary, and stellar quantities.

```python
feature_cols = [
    "pl_orbper",   # orbital period [days]
    "pl_orbsmax",  # semi-major axis [AU]
    "pl_orbeccen", # eccentricity
    "pl_bmasse",   # planet mass or M*sin(i) [Earth masses]
    "pl_dens",     # density [g/cm^3]
    "pl_eqt",      # equilibrium temperature [K]
    "pl_insol",    # insolation flux [Earth flux]
    "st_teff",     # stellar effective temperature [K]
    "st_rad",      # stellar radius [Solar radii]
    "st_mass",     # stellar mass [Solar masses]
    "st_met",      # metallicity [dex]
    "st_logg",     # stellar surface gravity
    "sy_dist",     # distance [pc]
    "sy_vmag",     # V magnitude
    "sy_kmag",     # K magnitude
]

target_col = "planet_class"

data = df[feature_cols + [target_col]].copy()
data.head()
```

## 5. Optional astrophysical filtering

Archive tables may include heterogeneous measurement quality and strong incompleteness.

For a pedagogical first pass, we keep the dataset fairly broad.  
But for a stricter scientific analysis, students may wish to require:
- non-null mass,
- non-null equilibrium temperature,
- bounded eccentricity,
- removal of extreme outliers.

Below we do only minimal sanity filtering.

```python
# Remove obviously problematic or non-physical values in a conservative way
data = data.copy()
data = data[(data["pl_orbper"].isna()) | (data["pl_orbper"] > 0)]
data = data[(data["pl_orbsmax"].isna()) | (data["pl_orbsmax"] > 0)]
data = data[(data["pl_bmasse"].isna()) | (data["pl_bmasse"] > 0)]
data = data[(data["st_mass"].isna()) | (data["st_mass"] > 0)]
data = data[(data["st_rad"].isna()) | (data["st_rad"] > 0)]
data = data[(data["sy_dist"].isna()) | (data["sy_dist"] > 0)]

print("Filtered shape:", data.shape)
data[target_col].value_counts()
```

## 7. Train-test split

We perform a stratified split so the class proportions are preserved.

```python
X = data[feature_cols]
y = data[target_col]

X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.25,
    random_state=42,
    stratify=y
)

print("Train shape:", X_train.shape)
print("Test shape:", X_test.shape)
print("\nTraining class distribution:")
print(y_train.value_counts(normalize=True))
print("\nTest class distribution:")
print(y_test.value_counts(normalize=True))
```

## 8. Build the preprocessing + model pipeline

Random Forest does **not** require feature scaling, but it does require a strategy for missing values.

We therefore use:
- median imputation for all numeric columns,
- RandomForestClassifier for the model.

We also use `class_weight="balanced_subsample"` to reduce the effect of class imbalance.

```python
numeric_features = feature_cols

numeric_transformer = Pipeline(steps=[
    ("imputer", SimpleImputer(strategy="median"))
])

preprocessor = ColumnTransformer(
    transformers=[
        ("num", numeric_transformer, numeric_features)
    ],
    remainder="drop"
)

rf = RandomForestClassifier(
    n_estimators=500,
    max_depth=None,
    min_samples_leaf=3,
    min_samples_split=6,
    max_features="sqrt",
    class_weight="balanced_subsample",
    random_state=42,
    n_jobs=-1
)

model = Pipeline(steps=[
    ("preprocessor", preprocessor),
    ("classifier", rf)
])

model
```

## 9. Cross-validation on the training set

We evaluate with:
- balanced accuracy,
- macro F1,
- weighted F1.

This is better than relying on one train/test split alone.

```python
cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)

cv_results = cross_validate(
    model,
    X_train,
    y_train,
    cv=cv,
    scoring={
        "balanced_accuracy": "balanced_accuracy",
        "f1_macro": "f1_macro",
        "f1_weighted": "f1_weighted"
    },
    n_jobs=-1,
    return_train_score=False
)

pd.DataFrame(cv_results)
```

```python
summary = pd.DataFrame({
    metric: [np.mean(values), np.std(values)]
    for metric, values in cv_results.items()
    if metric.startswith("test_")
}, index=["mean", "std"]).T

summary
```

### Student task
Interpret the difference between:
- macro F1
- weighted F1
- balanced accuracy

Which one is most appropriate if the rarest class is scientifically important?

## 10. Fit final model and evaluate on the test set

```python
model.fit(X_train, y_train)
y_pred = model.predict(X_test)

print("Balanced accuracy:", balanced_accuracy_score(y_test, y_pred))
print("Macro F1:", f1_score(y_test, y_pred, average="macro"))
print("Weighted F1:", f1_score(y_test, y_pred, average="weighted"))
print()
print(classification_report(y_test, y_pred))
```

```python
fig, ax = plt.subplots(figsize=(6,5))
ConfusionMatrixDisplay.from_predictions(y_test, y_pred, ax=ax, cmap="Blues", colorbar=False)
ax.set_title("Confusion matrix on held-out test set")
plt.show()
```

## 11. Permutation importance

Tree-based impurity importance can be biased, so we also compute **permutation importance** on the test set.

This is often more trustworthy for interpretation.

```python
perm = permutation_importance(
    model,
    X_test,
    y_test,
    n_repeats=10,
    random_state=42,
    scoring="balanced_accuracy",
    n_jobs=-1
)

importance_df = pd.DataFrame({
    "feature": feature_cols,
    "importance_mean": perm.importances_mean,
    "importance_std": perm.importances_std
}).sort_values("importance_mean", ascending=False)

importance_df
```

```python
plt.figure(figsize=(8,6))
plt.barh(importance_df["feature"], importance_df["importance_mean"], xerr=importance_df["importance_std"])
plt.gca().invert_yaxis()
plt.xlabel("Permutation importance (decrease in balanced accuracy)")
plt.title("Feature importance on held-out test set")
plt.show()
```

## 12. Optional extension: compare two feature sets

A useful scientific exercise is to compare:

### Model A
All selected features

### Model B
Only stellar + orbital features  
(excluding direct planet-structure features such as `pl_bmasse` and `pl_dens`)

This tests how much of the classification depends on direct structural information versus environment.

```python
feature_cols_B = [
    "pl_orbper",
    "pl_orbsmax",
    "pl_orbeccen",
    "pl_eqt",
    "pl_insol",
    "st_teff",
    "st_rad",
    "st_mass",
    "st_met",
    "st_logg",
    "sy_dist",
    "sy_vmag",
    "sy_kmag",
]

X_B = data[feature_cols_B]
X_train_B, X_test_B, y_train_B, y_test_B = train_test_split(
    X_B, y,
    test_size=0.25,
    random_state=42,
    stratify=y
)

preprocessor_B = ColumnTransformer(
    transformers=[
        ("num", Pipeline([("imputer", SimpleImputer(strategy="median"))]), feature_cols_B)
    ]
)

model_B = Pipeline(steps=[
    ("preprocessor", preprocessor_B),
    ("classifier", RandomForestClassifier(
        n_estimators=500,
        min_samples_leaf=3,
        min_samples_split=6,
        max_features="sqrt",
        class_weight="balanced_subsample",
        random_state=42,
        n_jobs=-1
    ))
])

cv_results_B = cross_validate(
    model_B,
    X_train_B,
    y_train_B,
    cv=cv,
    scoring={
        "balanced_accuracy": "balanced_accuracy",
        "f1_macro": "f1_macro",
        "f1_weighted": "f1_weighted"
    },
    n_jobs=-1
)

summary_B = pd.DataFrame({
    metric: [np.mean(values), np.std(values)]
    for metric, values in cv_results_B.items()
    if metric.startswith("test_")
}, index=["mean", "std"]).T

summary_B
```

### Compare the two models
If Model A substantially outperforms Model B, that likely means direct planet-structure features carry much of the predictive signal. That is not surprising, but it is scientifically useful to quantify.

This comparison also opens a very good classroom discussion:
> Is the model learning a physically deep relation, or is it largely exploiting nearby observables?

We can see that model A outperforms B

## 13. Optional extension: regression instead of classification

A strong extension for advanced students is to switch from classification to regression and predict:
- `pl_bmasse`, or
- `pl_eqt`, or
- `pl_dens`

using `RandomForestRegressor`.

Students can then compare:
- RMSE / MAE,
- residual plots,
- feature importance,
- regression versus classification framing.

```python
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score
from sklearn.model_selection import KFold

# ── Target: planet mass [Earth masses] ────────────────────────────────────────
# pl_rade is re-included: it is no longer a leakage risk because the target is
# now mass, not the radius class.  pl_dens is excluded — density encodes
# mass/radius³ directly and would make the task trivially easy.
reg_target = "pl_bmasse"
reg_features = ["pl_rade"] + [f for f in feature_cols if f not in ["pl_bmasse", "pl_dens"]]

reg_data = df[reg_features + [reg_target]].copy()
reg_data = reg_data[reg_data[reg_target].notna() & (reg_data[reg_target] > 0)].copy()

# Log-transform: pl_bmasse spans ~4 orders of magnitude (0.02 – 9500 M_Earth)
reg_data["log10_bmasse"] = np.log10(reg_data[reg_target])
print(f"Regression dataset size : {len(reg_data)}")
print(f"log10(pl_bmasse)  mean={reg_data['log10_bmasse'].mean():.2f}  "
      f"std={reg_data['log10_bmasse'].std():.2f}")

fig, ax = plt.subplots(1, 2, figsize=(10, 3))
ax[0].hist(reg_data[reg_target], bins=60)
ax[0].set_xlabel("pl_bmasse [M_Earth]"); ax[0].set_title("Raw (linear)")
ax[1].hist(reg_data["log10_bmasse"], bins=60)
ax[1].set_xlabel("log10(pl_bmasse / M_Earth)"); ax[1].set_title("Log-transformed")
plt.suptitle("Distribution of the regression target"); plt.tight_layout(); plt.show()

# ── Train / test split ────────────────────────────────────────────────────────
X_reg = reg_data[reg_features]
y_reg = reg_data["log10_bmasse"]

X_reg_train, X_reg_test, y_reg_train, y_reg_test = train_test_split(
    X_reg, y_reg, test_size=0.25, random_state=42
)

# ── Pipeline ──────────────────────────────────────────────────────────────────
reg_preprocessor = ColumnTransformer([
    ("num", Pipeline([("imputer", SimpleImputer(strategy="median"))]), reg_features)
])

reg_model = Pipeline([
    ("preprocessor", reg_preprocessor),
    ("regressor", RandomForestRegressor(
        n_estimators=500,
        min_samples_leaf=3,
        min_samples_split=6,
        max_features="sqrt",
        random_state=42,
        n_jobs=-1
    ))
])

# ── 5-fold cross-validation on training set ───────────────────────────────────
cv_reg = KFold(n_splits=5, shuffle=True, random_state=42)
cv_reg_res = cross_validate(
    reg_model, X_reg_train, y_reg_train, cv=cv_reg,
    scoring={"r2": "r2",
             "neg_rmse": "neg_root_mean_squared_error",
             "neg_mae":  "neg_mean_absolute_error"},
    n_jobs=-1
)

print("\n5-fold CV on training set (target: log10(pl_bmasse))")
print(f"  R²   = {np.mean(cv_reg_res['test_r2']):.4f} ± {np.std(cv_reg_res['test_r2']):.4f}")
print(f"  RMSE = {-np.mean(cv_reg_res['test_neg_rmse']):.4f} ± {np.std(cv_reg_res['test_neg_rmse']):.4f} dex")
print(f"  MAE  = {-np.mean(cv_reg_res['test_neg_mae']):.4f} ± {np.std(cv_reg_res['test_neg_mae']):.4f} dex")
```

```python
# ── Fit on full training set and evaluate on test set ─────────────────────────
reg_model.fit(X_reg_train, y_reg_train)
y_reg_pred = reg_model.predict(X_reg_test)

rmse = np.sqrt(mean_squared_error(y_reg_test, y_reg_pred))
mae  = mean_absolute_error(y_reg_test, y_reg_pred)
r2   = r2_score(y_reg_test, y_reg_pred)

print("Test-set metrics (target: log10(pl_bmasse))")
print(f"  R²   = {r2:.4f}")
print(f"  RMSE = {rmse:.4f} dex  (~factor {10**rmse:.2f}× in linear mass)")
print(f"  MAE  = {mae:.4f} dex  (~factor {10**mae:.2f}× in linear mass)")

# ── Diagnostic plots ──────────────────────────────────────────────────────────
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

axes[0].scatter(y_reg_test, y_reg_pred, s=8, alpha=0.35)
lo = min(y_reg_test.min(), y_reg_pred.min())
hi = max(y_reg_test.max(), y_reg_pred.max())
axes[0].plot([lo, hi], [lo, hi], "r--", lw=1)
axes[0].set_xlabel(r"True  $\log_{10}$(M_p / M_Earth)")
axes[0].set_ylabel(r"Predicted  $\log_{10}$(M_p / M_Earth)")
axes[0].set_title(fr"Predicted vs True  ($R^2$ = {r2:.3f})")

residuals = y_reg_pred - y_reg_test
axes[1].scatter(y_reg_pred, residuals, s=8, alpha=0.35)
axes[1].axhline(0, color="r", lw=1, ls="--")
axes[1].set_xlabel(r"Predicted  $\log_{10}$(M_p / M_Earth)")
axes[1].set_ylabel("Residual  (pred - true) [dex]")
axes[1].set_title("Residual plot")

plt.suptitle("Random Forest Regression - planet mass")
plt.tight_layout()
plt.show()

# Residual distribution
plt.figure(figsize=(6, 3))
plt.hist(residuals, bins=50)
plt.xlabel("Residual [dex]")
plt.title(f"Residual distribution  (mean={residuals.mean():.3f}, std={residuals.std():.3f})")
plt.show()
```

```python
# ── Permutation importance for the regression model ───────────────────────────
perm_reg = permutation_importance(
    reg_model, X_reg_test, y_reg_test,
    n_repeats=10, random_state=42, scoring="r2", n_jobs=-1
)

perm_reg_df = pd.DataFrame({
    "feature":          reg_features,
    "importance_mean":  perm_reg.importances_mean,
    "importance_std":   perm_reg.importances_std,
}).sort_values("importance_mean", ascending=False)

plt.figure(figsize=(8, 6))
plt.barh(perm_reg_df["feature"], perm_reg_df["importance_mean"],
         xerr=perm_reg_df["importance_std"])
plt.gca().invert_yaxis()
plt.xlabel(r"Permutation importance (decrease in $R^2$)")
plt.title(r"Feature importance: predicting $\log_{10}$(planet mass)")
plt.tight_layout()
plt.show()

print(perm_reg_df.to_string(index=False))
```

Physics interpretation:
- pl_rade dominates via the mass-radius relation ($M \propto R^{~3.7}$ for rocky, ~2 for gas giants).
- pl_orbsmax carries the next-strongest signal: more massive planets are found at wider orbits (detection bias + disk formation physics).
- Stellar features (st_mass, st_met) contribute via the planet-formation connection: more massive / metal-rich stars preferentially host more massive planets.")

### Secondary target: equilibrium temperature (`pl_eqt`)

`pl_eqt` has fewer measurements (~75% coverage) but is astrophysically fundamental: it sets the planet's energy balance and is relevant to habitability studies. The expected dominant feature is `pl_orbsmax` via $T_{\rm eq} \propto T_\star (R_\star / a)^{1/2}$, making this a physically cleaner regression task than mass prediction.

```python
# ── Secondary regression: equilibrium temperature ─────────────────────────────
# pl_rade is included; pl_insol is excluded (it's a direct proxy for T_eq via
# S ∝ L/a²  and  T_eq ∝ S^0.25).
eqt_target   = "pl_eqt"
eqt_features = ["pl_rade"] + [f for f in feature_cols if f not in ["pl_eqt", "pl_insol"]]

eqt_data = df[eqt_features + [eqt_target]].copy()
eqt_data = eqt_data[eqt_data[eqt_target].notna() & (eqt_data[eqt_target] > 0)].copy()
print(f"pl_eqt regression dataset size: {len(eqt_data)}")

X_eqt = eqt_data[eqt_features]
y_eqt = eqt_data[eqt_target]          # predict T_eq in Kelvin (linear is fine here)

X_eqt_train, X_eqt_test, y_eqt_train, y_eqt_test = train_test_split(
    X_eqt, y_eqt, test_size=0.25, random_state=42
)

eqt_preprocessor = ColumnTransformer([
    ("num", Pipeline([("imputer", SimpleImputer(strategy="median"))]), eqt_features)
])

eqt_model = Pipeline([
    ("preprocessor", eqt_preprocessor),
    ("regressor", RandomForestRegressor(
        n_estimators=500, min_samples_leaf=3, min_samples_split=6,
        max_features="sqrt", random_state=42, n_jobs=-1
    ))
])

eqt_model.fit(X_eqt_train, y_eqt_train)
y_eqt_pred = eqt_model.predict(X_eqt_test)

eqt_rmse = np.sqrt(mean_squared_error(y_eqt_test, y_eqt_pred))
eqt_mae  = mean_absolute_error(y_eqt_test, y_eqt_pred)
eqt_r2   = r2_score(y_eqt_test, y_eqt_pred)

print(f"\nTest-set metrics (target: pl_eqt [K])")
print(f"  R²   = {eqt_r2:.4f}")
print(f"  RMSE = {eqt_rmse:.1f} K")
print(f"  MAE  = {eqt_mae:.1f} K")

fig, axes = plt.subplots(1, 2, figsize=(12, 5))
axes[0].scatter(y_eqt_test, y_eqt_pred, s=8, alpha=0.35)
lo, hi = y_eqt_test.min(), y_eqt_test.max()
axes[0].plot([lo, hi], [lo, hi], "r--", lw=1)
axes[0].set_xlabel("True T_eq [K]"); axes[0].set_ylabel("Predicted T_eq [K]")
axes[0].set_title(f"Predicted vs True  ($R^2$ = {eqt_r2:.3f})")

eqt_res = y_eqt_pred - y_eqt_test
axes[1].scatter(y_eqt_pred, eqt_res, s=8, alpha=0.35)
axes[1].axhline(0, color="r", lw=1, ls="--")
axes[1].set_xlabel("Predicted T_eq [K]"); axes[1].set_ylabel("Residual [K]")
axes[1].set_title("Residual plot")
plt.suptitle("Random Forest Regression - equilibrium temperature")
plt.tight_layout(); plt.show()

# Importance
perm_eqt = permutation_importance(
    eqt_model, X_eqt_test, y_eqt_test,
    n_repeats=10, random_state=42, scoring="r2", n_jobs=-1
)
perm_eqt_df = pd.DataFrame({
    "feature": eqt_features,
    "importance_mean": perm_eqt.importances_mean,
    "importance_std":  perm_eqt.importances_std,
}).sort_values("importance_mean", ascending=False)

plt.figure(figsize=(8, 6))
plt.barh(perm_eqt_df["feature"], perm_eqt_df["importance_mean"],
         xerr=perm_eqt_df["importance_std"])
plt.gca().invert_yaxis()
plt.xlabel("Permutation importance (decrease in R²)")
plt.title("Feature importance: predicting T_eq [K]")
plt.tight_layout(); plt.show()

print(perm_eqt_df.to_string(index=False))
print()
print("Physics check: pl_orbsmax and st_teff / st_rad should dominate")
print("because T_eq ∝ T_eff * sqrt(R_star / a) from radiative equilibrium.")
```

### Regression vs classification: a comparison

| Aspect | Classification (radius class) | Regression (mass / T_eq) |
|---|---|---|
| **Target** | Discrete label (rocky / sub-Neptune / giant) | Continuous value (M_Earth or K) |
| **Metric** | Balanced accuracy, macro-F1 | R², RMSE, MAE |
| **Leakage risk** | `pl_rade` defines the class → must exclude | Exclude the target itself and any algebraic proxies (e.g. `pl_dens` for mass) |
| **Interpretability** | Confusion matrix reveals which boundaries are hard | Residual plot reveals where errors concentrate |
| **Feature importance** | `pl_bmasse` and `pl_dens` dominate (structural) | For mass: `pl_rade` dominates (M-R relation); for T_eq: `pl_orbsmax` and `st_teff` dominate (radiative equilibrium) |
| **Scientific insight** | Tests whether orbital/stellar context predicts compositional class | Tests whether the model has learnt a physical scaling law |

**Key lesson:** the regression task for `pl_eqt` is close to fitting the analytical formula $T_{\rm eq} \propto T_\star (R_\star/a)^{1/2}$. A high R² there confirms the model has encoded this physics — but it does not mean it *understands* it.

## 14. Scientific caveats students must mention in their report

Any serious report should discuss the following:

### 1. Archive heterogeneity
The confirmed exoplanet sample is not a homogeneous survey product.

### 2. Selection effects
Transit, radial velocity, imaging, microlensing, and timing methods probe different regions of parameter space.

### 3. Missing-data mechanism
Missingness is not necessarily random.

### 4. Target construction
The class boundaries (`1.8` and `4.0` Earth radii) are pedagogical approximations.

### 5. Feature leakage risk
Always verify that target-defining quantities are not indirectly reintroduced.

### 6. Prediction is not causation
A strong classifier does not prove that the top feature is a causal driver.
