# First Dataset Validation

This notebook shows the exploration of the dataset obtained from the first test of the automated pipeline. 10 LC per class was generated as described in Phase2.md

```python
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from pathlib import Path

plt.style.use("seaborn-v0_8-deep")

DATASET_DIR = Path("../data/lightcurves/") # dir holding the .dat light curves
META_PATH   = Path("../data/metadata.csv") # metadata file
```

## Read data

```python
# find all .dat files in the lightcurves directory
dat_files = [f for f in os.listdir(DATASET_DIR) if f.endswith(".dat")]
print(f"Found {len(dat_files)} .dat files:")
```

```python
df = pd.read_csv(META_PATH, delimiter=",", dtype={"file_id": str,"activity_class": str})
```

```python
labels = {0: "0: mild, no planet", 1: "1: strong, no planet",
          2: "2: mild, planet",   3: "3: strong, planet"}

df["label_name"] = df["label"].map(labels)
```

```python
print(df.columns)
```

```python
def load_lc(file_id):
    """Load a PSLS .dat LC. Returns (time_days, flux_ppm)"""
    p = DATASET_DIR / f"{file_id}.dat"
    arr = np.loadtxt(p, comments=("#"))
    t, f = arr[:, 0], arr[:, 1]
    t = t / 86400.0
    return t, f
```

## Example lightcurves per class

```python
rows = []
for fid in df["file_id"]:
    t, f = load_lc(fid)
    rows.append(dict(
        file_id=fid,
        npoints=t.size,
        rms_ppm=round(np.std(f), 1),
    ))
sig = pd.DataFrame(rows)
df = df.merge(sig, on="file_id", how="left")

print("\nrms_ppm by class:")
print(df.groupby("label")["rms_ppm"].describe()[["mean", "min", "max"]].round(1))
# strong classes (1,3) should have HIGHER rms than mild (0,2)
```

```python
# Plot 1 light curve file for each class
classes = df['label'].unique()

sample_files = []
for cls in classes:
    cls_files = df[df['label'] == cls]['file_id'].values
    if len(cls_files) > 0:
        sample_files.append(f"{cls_files[0]}")  # Take the first file for this class
sample_files.sort() 

for dat_file in sample_files:
    plt.figure(figsize=(16, 5))
    t, f = load_lc(dat_file)
    f = f / 1e4  # Convert from ppm to relative flux
    class_label = df.loc[df['file_id'] == dat_file, 'label_name'].values[0]
    plt.plot(t, f)
    plt.xlabel("Time [days]")
    plt.ylabel("Flux [relative %]")
    plt.title(f"{dat_file} psls LC (Class {class_label})")
    #plt.savefig(f"lightcurve_{file_id}.png")
    plt.show()
    plt.close()
```

## Validation

```python
def folded_depth_ppm(t, f, period, n_bins=200):
    """Crude depth: median out-of-transit minus the deepest phase bin."""
    phase = (t % period) / period
    idx = np.argsort(phase)
    ph, fl = phase[idx], f[idx]
    edges = np.linspace(0, 1, n_bins + 1)
    binned = np.array([np.median(fl[(ph >= edges[i]) & (ph < edges[i+1])])
                       if ((ph >= edges[i]) & (ph < edges[i+1])).any() else np.nan
                       for i in range(n_bins)])
    base = np.nanmedian(binned)
    deepest = np.nanmin(binned)
    return (base - deepest)

planets = df[df["planet_present"] == 1]
rec = []
for _, r in planets.iterrows():
    t, f = load_lc(r["file_id"])
    meas = folded_depth_ppm(t, f, r["planet_period_days"])
    rec.append(dict(file_id=r["file_id"], label=r["label"],
                    injected_ppm=round(r["transit_depth_ppm"], 1),
                    measured_ppm=round(meas, 1),
                    ratio=round(meas / r["transit_depth_ppm"], 2)))
depth = pd.DataFrame(rec)
print(depth.to_string(index=False))
print("\nmedian measured/injected ratio:", round(depth["ratio"].median(), 2),
      "(want ~1; activity inflates the crude estimator so >1 is expected)")
```

```python
from astropy.timeseries import LombScargle

active = df[df["activity_class"] == "strong"]
prot_rec = []
for _, r in active.iterrows():
    t, f = load_lc(r["file_id"])
    # search 2–60 d
    freq = np.linspace(1/60, 1/2, 5000)
    power = LombScargle(t, f).power(freq)
    p_peak = 1 / freq[np.argmax(power)]
    prot_rec.append(dict(file_id=r["file_id"],
                         injected_prot=r["rotation_period_days"],
                         peak_prot=round(p_peak, 2)))
prot = pd.DataFrame(prot_rec)
prot["close"] = (abs(prot["peak_prot"] - prot["injected_prot"])
                 / prot["injected_prot"] < 0.2)
print(prot.to_string(index=False))
print(f"\nrecovered within 20%: {prot['close'].mean():.0%} "
      "(won't be 100% — flares/multi-spot/harmonics shift the peak)")
```

```python
import matplotlib.pyplot as plt

check_cols = ["teff", "logg", "rotation_period_days", "sigma_ppm", "rms_ppm"]
fig, axes = plt.subplots(1, len(check_cols), figsize=(4*len(check_cols), 3.5))
for ax, col in zip(axes, check_cols):
    for lab, g in df.groupby("label"):
        ax.hist(g[col].dropna(), bins=20, alpha=0.45, label=f"C{lab}")
    ax.set_title(col); ax.legend(fontsize=7)
plt.tight_layout(); plt.show()

# Numeric leakage test: planet-present vs absent should match on stellar params
print("Planet vs no-planet — mean stellar params (should be ~equal):")
print(df.groupby("planet_present")[["teff", "logg"]].mean().round(1))

# Duplicate-realization check (MasterSeed must vary)
dups = df["seed"].duplicated().sum()
print(f"\nduplicate seeds: {dups} (must be 0)")
fingerprints = {fid: round(np.std(load_lc(fid)[1])*1e9) for fid in df["file_id"]}
print("identical-rms collisions:",
      len(fingerprints) - len(set(fingerprints.values())))
```

## Transit recovery

```python
from astropy.timeseries import BoxLeastSquares
from wotan import flatten

def bin_lc(t, f, bin_width=1/24):
    """Average into fixed-width time bins (default 1 h = 1/24 d)."""
    bins = np.arange(t.min(), t.max() + bin_width, bin_width)
    idx  = np.digitize(t, bins)
    t_b, f_b = [], []
    for i in range(1, len(bins)):
        m = idx == i
        if m.sum() > 0:
            t_b.append(t[m].mean())
            f_b.append(f[m].mean())
    return np.array(t_b), np.array(f_b)

durations    = np.array([0.05, 0.10, 0.16, 0.25])   # candidate transit durations [days]
planet_rows  = df[df["planet_present"] == 1]
bls_rec      = []

for _, r in planet_rows.iterrows():
    t, f = load_lc(r["file_id"])

    # 1. Wotan biweight detrend (window 0.5 d)
    #    Convert ppm → normalised flux so Wotan doesn't divide by ~0
    f_norm        = 1.0 + f / 1e4
    flat, _       = flatten(t, f_norm, method="biweight", window_length=0.5, return_trend=True)
    f_det         = (flat - 1.0) * 1e4      # detrended flux back in ppm

    # 2. Hour-bin the detrended native-cadence flux
    t_bin, f_bin  = bin_lc(t, f_det)

    # 3. BLS on the binned data
    model         = BoxLeastSquares(t_bin, f_bin)
    results       = model.autopower(durations, minimum_period=1.0, maximum_period=50.0)

    best_idx      = np.argmax(results.power)
    best_period   = float(results.period[best_idx])
    best_duration = float(results.duration[best_idx])
    best_t0       = float(results.transit_time[best_idx])

    # 4. Depth from BLS box-model fit (not deepest phase bin)
    stats         = model.compute_stats(best_period, best_duration, best_t0)
    bls_depth     = abs(float(stats["depth"][0]))
    bls_depth_err = abs(float(stats["depth"][1]))

    period_err_pct = abs(best_period - r["planet_period_days"]) / r["planet_period_days"] * 100

    bls_rec.append(dict(
        file_id=r["file_id"],
        label=r["label"],
        injected_period=round(r["planet_period_days"], 4),
        bls_period=round(best_period, 4),
        period_err_pct=round(period_err_pct, 1),
        injected_depth_ppm=round(r["transit_depth_ppm"], 1),
        bls_depth_ppm=round(bls_depth, 1),
        depth_err_ppm=round(bls_depth_err, 1),
    ))

    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 4))

    ax1.plot(results.period, results.power, lw=0.7)
    ax1.axvline(r["planet_period_days"], color="red", ls="--",
                label=f"injected {r['planet_period_days']:.4f} d")
    ax1.axvline(best_period, color="orange", ls=":",
                label=f"BLS {best_period:.4f} d")
    ax1.set_xlabel("Period [days]")
    ax1.set_ylabel("BLS power")
    ax1.set_title(f"{r['file_id']}  (class {r['label']})")
    ax1.legend(fontsize=8)

    phase    = ((t_bin - best_t0) % best_period) / best_period
    phase[phase > 0.5] -= 1.0
    sort_idx = np.argsort(phase)
    ax2.plot(phase[sort_idx], f_bin[sort_idx], ".", ms=2, alpha=0.6, color="steelblue")
    ax2.axhline(0, color="gray", lw=0.5, ls=":")
    ax2.set_xlabel("Phase")
    ax2.set_ylabel("Detrended flux [ppm]")
    ax2.set_title(f"P = {best_period:.4f} d  |  depth = {bls_depth:.0f} ± {bls_depth_err:.0f} ppm")
    ax2.set_xlim(-0.5, 0.5)

    plt.tight_layout()
    plt.show()

bls_df = pd.DataFrame(bls_rec)
bls_df["recovered"]   = bls_df["period_err_pct"] < 10.0
bls_df["depth_ratio"] = (bls_df["bls_depth_ppm"] / bls_df["injected_depth_ppm"]).round(2)

print("=== Per-curve BLS recovery ===")
print(bls_df.drop(columns="depth_ratio").to_string(index=False))

print("\n=== Stratified recovery by activity class ===")
summary = bls_df.groupby("label").agg(
    n=("file_id", "count"),
    pct_recovered=("recovered", lambda x: f"{x.mean():.0%}"),
    med_period_err_pct=("period_err_pct", "median"),
    med_depth_ratio=("depth_ratio", "median"),
)
summary.insert(0, "class", summary.index.map(labels))
print(summary.to_string())
```
